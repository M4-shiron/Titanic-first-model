```python
# 1. PyTorch（パイトーチ）のツール箱を読み込む
import torch

# 2. まずは普通のPythonの「数字のリスト」を作る
my_list = [1, 2, 3, 4, 5]
print("普通のリスト:", my_list)

# 3. リストを「テンソル」に変身させる！
my_tensor = torch.tensor(my_list)
print("テンソルに変身:", my_tensor)

# 4. テンソルならではの「一括計算（行列計算）」を体験してみる
# ディープラーニングでは、何万個ものデータをいっぺんに掛け算します
result = my_tensor * 10
print("全部いっぺんに10倍:", result)
```

```python
import torch.nn as nn

# 1. 「1つの手がかり」から「1つの答え」を出す、一番小さな脳細胞（AI）を作る
simple_ai = nn.Linear(in_features=1, out_features=1)

# 2. 手がかりとなるテンソルを作る（例：年齢が28歳だとします）
# ※AIが計算しやすいように、少数を表す「.0」をつけています
input_tensor = torch.tensor([28.0])

# 3. AIに年齢を入力して、予測を出させてみる！
prediction = simple_ai(input_tensor)

print("AIの出力結果:", prediction)
```

```python
import torch.optim as optim

# 1. 練習問題と、本当の答えを用意する
# 先ほどの simple_ai をそのまま使います
input_tensor = torch.tensor([28.0]) # 入力：28歳
target_tensor = torch.tensor([1.0]) # 正解：1（生存したとする）

# 2. 「ズレを計算するツール（MSELoss）」と「AIを修正するツール（SGD）」を準備
criterion = nn.MSELoss()
optimizer = optim.SGD(simple_ai.parameters(), lr=0.001) # lrは「1歩の歩幅」

# --- ここからが「1回の学習」のステップ ---
optimizer.zero_grad() # 前の反省を一旦リセット

# ① 予測する
prediction = simple_ai(input_tensor)

# ② 答え合わせ（正解とのズレ＝誤差 を計算）
loss = criterion(prediction, target_tensor)
print(f"修正前の誤差（ズレ）: {loss.item():.4f}")

# ③ 誤差から「どう直すべきか」を計算（自動微分）
loss.backward()

# ④ 実際にAIの脳内を修正する！（谷を1歩下る）
optimizer.step()

# もう一度予測して、誤差が減ったか確認
new_prediction = simple_ai(input_tensor)
new_loss = criterion(new_prediction, target_tensor)
print(f"1回学習した後の誤差: {new_loss.item():.4f}")
```

```python
# 先ほどと同じように、新しく「生まれたてのAI」を準備
simple_ai = nn.Linear(in_features=1, out_features=1)
optimizer = optim.SGD(simple_ai.parameters(), lr=0.001)

print("--- 100回の猛特訓スタート！ ---")

# for文を使って、さっきの学習ステップを100回（100エポック）繰り返す！
for epoch in range(100):
    optimizer.zero_grad()           # ① 前の反省をリセット
    prediction = simple_ai(input_tensor) # ② 予測する
    loss = criterion(prediction, target_tensor) # ③ 答え合わせ（誤差の計算）
    loss.backward()                 # ④ どう直すべきか計算
    optimizer.step()                # ⑤ 実際に脳内を修正！（1歩下る）
    
    # 10回ごとに、誤差がどう減っているか報告させる
    if (epoch + 1) % 10 == 0:
        print(f"{epoch + 1}回目: 誤差 = {loss.item():.4f}")

# 特訓が終わった後のAIに、もう一度答えを出させてみる
final_prediction = simple_ai(input_tensor)
print(f"\n★ 特訓後のAIの予測値（目標は1.0）: {final_prediction.item():.4f}")
```
